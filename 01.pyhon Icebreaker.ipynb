```python
import numpy as np

import matplotlib.pyplot as plt

# Function to compute the squared sum of two numbers


def squared_sum(x, y):
    return x**2 + y**2

# Function to return 'YES' if (A and B) or (C xor A) and 'NO' otherwise


def ifs_one(A, B, C):
    if (A and B) or (C ^ A):
        return 'YES'
    else:
        return 'NO'

# Function to generate the sequence 2*i + (i-1) from 2 to N, if N is greater than 2. Return 0 otherwise


def loops_one(N):
    if N > 2:
        return [2*i + (i-1) for i in range(2, N+1)]
    else:
        return 0

# Function to plot sin(x) between 0 and 2π


def plot_sin():
    x = np.linspace(0, 2*np.pi, 100)
    y = np.sin(x)
    plt.plot(x, y)
    plt.xlabel('x')
    plt.ylabel('sin(x)')
    plt.title('Plot of sin(x) between 0 and 2π')
    plt.grid(True)
    plt.show()


# Example usage
a = 3
b = 4
c = 5
n = 5

print("Squared sum of a and b:", squared_sum(a, b))
print("ifs_one result:", ifs_one(a, b, c))
print("loops_one result:", loops_one(n))
plot_sin()
```
